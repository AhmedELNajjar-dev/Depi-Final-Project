## Cleaning the tuesday's data

> removing the spaces in the columns name

> Dropping the null values as it is less than 0.045%

> removing duplicated values the null values

```python
tday.columns=tday.columns.str.strip()
tday.head()
```

#### drop nulls

```python
tday.dropna(inplace=True,axis=1)
```

```python
tday.drop_duplicates(inplace=True)
tday.duplicated().sum()
```
